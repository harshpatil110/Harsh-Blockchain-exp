````python
# Import required libraries

import datetime
import hashlib
import json

print("Required libraries imported successfully.")
````

## Step 1: Create the Blockchain Class

The Blockchain class stores:
- The complete chain of blocks.
- Pending transactions waiting to be mined.
- The Genesis Block, which is the first block in the Blockchain.

````python
class Blockchain:

    def __init__(self):
        # Store the complete blockchain
        self.chain = []

        # Store pending transactions
        self.transactions = []

        # Create the Genesis Block
        self.create_block(
            proof=1,
            previous_hash='0',
            transactions=[]
        )

    def create_block(self, proof, previous_hash, transactions):
        # Create a new block
        block = {
            'index': len(self.chain) + 1,
            'timestamp': str(datetime.datetime.now()),
            'transactions': transactions,
            'proof': proof,
            'previous_hash': previous_hash
        }

        # Add the block to the blockchain
        self.chain.append(block)

        return block

    def get_previous_block(self):
        # Return the latest block
        return self.chain[-1]

    def create_transactions(self, sender, receiver, amount):
        # Create a transaction
        transaction = {
            'sender': sender,
            'receiver': receiver,
            'amount': amount
        }

        # Add transaction to pending transaction list
        self.transactions.append(transaction)

        # Return index of the future block
        return self.get_previous_block()['index'] + 1

    def proof_of_work(self, previous_proof):
        # Search for the Golden Nonce
        new_proof = 1
        check_proof = False

        while check_proof is False:
            # Cryptographic puzzle
            value = str(new_proof**2 - previous_proof**2)

            hash_operation = hashlib.sha256(
                value.encode()
            ).hexdigest()

            # Required: hash should start with 000
            if hash_operation[:3] == '000':
                check_proof = True
            else:
                new_proof += 1

        return new_proof

    def hash(self, block):
        # Convert block into JSON with sorted keys
        encoded_block = json.dumps(
            block,
            sort_keys=True
        ).encode()

        # Generate SHA-256 hash
        return hashlib.sha256(encoded_block).hexdigest()

    def is_chain_valid(self, chain):
        previous_block = chain[0]
        block_index = 1

        while block_index < len(chain):
            block = chain[block_index]

            # Check previous hash
            if block['previous_hash'] != self.hash(previous_block):
                return False

            # Check Proof of Work
            previous_proof = previous_block['proof']
            proof = block['proof']

            value = str(proof**2 - previous_proof**2)

            hash_operation = hashlib.sha256(
                value.encode()
            ).hexdigest()

            # The hash must start with 000
            if hash_operation[:3] != '000':
                return False

            previous_block = block
            block_index += 1

        return True
````

## Step 2: Create the Blockchain

An object of the `Blockchain` class is created. This automatically creates the Genesis Block.

````python
blockchain = Blockchain()

print("Blockchain created successfully.")
print("\nGenesis Block:")
print(blockchain.chain[0])
````

## Step 3: Create Demo Transactions



````python
blockchain.create_transactions(
    sender='Harsh',
    receiver='Rohit',
    amount=100
)

blockchain.create_transactions(
    sender='Rohit',
    receiver='Chinmay',
    amount=50
)

blockchain.create_transactions(
    sender='Chinmay',
    receiver='Amit',
    amount=75
)

blockchain.create_transactions(
    sender='Amit',
    receiver='Rahul',
    amount=25
)

print("Transactions created successfully.")
````

## Step 4: Display Pending Transactions

````python
print("Pending Transactions:")

for transaction in blockchain.transactions:
    print(transaction)
````

## Step 5: Mine the Block

According to the experiment:
1. Mine only when transactions are available.
2. Search for the Golden Nonce.
3. The SHA-256 hash must start with `"000"`.
4. Copy the transactions for the new block.
5. Remove the transactions from the pending list before creating the block.
6. Create the new block.

````python
if len(blockchain.transactions) == 0:

    print("No transactions available for mining.")

else:

    # Get the previous block
    previous_block = blockchain.get_previous_block()

    # Get previous proof
    previous_proof = previous_block['proof']

    print("Mining started...")
    print("Searching for Golden Nonce...")

    # Find Golden Nonce
    proof = blockchain.proof_of_work(previous_proof)

    # Calculate hash of previous block
    previous_hash = blockchain.hash(previous_block)

    # Copy transactions that will be mined
    transactions_to_mine = blockchain.transactions.copy()

    # Remove transactions from pending list before mining
    blockchain.transactions = []

    # Create the new block
    block = blockchain.create_block(
        proof=proof,
        previous_hash=previous_hash,
        transactions=transactions_to_mine
    )

    print("Block mined successfully!")
    print("\nNew Block:")
    print(json.dumps(block, indent=4))
````

## Step 6: Verify That Pending Transactions Were Removed

After mining, the pending transaction list should be empty because all transactions have been moved into the newly mined block.

````python
print("Pending Transactions After Mining:")
print(blockchain.transactions)

if len(blockchain.transactions) == 0:
    print("\nTransaction list is empty. Transactions were successfully removed.")
else:
    print("\nSome transactions are still pending.")
````

## Step 7: Add More Transactions

A Blockchain can continue accepting new transactions after a block has been mined.

````python
blockchain.create_transactions(
    sender='Harsh',
    receiver='Chinmay',
    amount=200
)

blockchain.create_transactions(
    sender='Rahul',
    receiver='Harsh',
    amount=125
)

print("New transactions added:")
for transaction in blockchain.transactions:
    print(transaction)
````

## Step 8: Mine the Second Transaction Block

````python
if len(blockchain.transactions) == 0:

    print("No transactions available for mining.")

else:

    previous_block = blockchain.get_previous_block()
    previous_proof = previous_block['proof']

    print("Mining second transaction block...")

    proof = blockchain.proof_of_work(previous_proof)
    previous_hash = blockchain.hash(previous_block)

    transactions_to_mine = blockchain.transactions.copy()

    # Clear pending transactions before block creation
    blockchain.transactions = []

    block = blockchain.create_block(
        proof=proof,
        previous_hash=previous_hash,
        transactions=transactions_to_mine
    )

    print("Second transaction block mined successfully!")
    print(json.dumps(block, indent=4))
````

## Step 9: Display the Complete Blockchain

````python
print("COMPLETE BLOCKCHAIN")
print("=" * 60)

for block in blockchain.chain:
    print(json.dumps(block, indent=4))
    print("-" * 60)

print("Total Number of Blocks:", len(blockchain.chain))
````

## Step 10: Check Blockchain Validity



````python
if blockchain.is_chain_valid(blockchain.chain):
    print("All good. The Blockchain is valid.")
else:
    print("The Blockchain is not valid.")
````

## Step 11: Demonstrate Blockchain Tampering Detection

The following cell creates a copy of the Blockchain and changes transaction data. The original Blockchain remains unchanged.

Run this cell to see that even a small modification makes the Blockchain invalid.

````python
# Create a deep copy of the blockchain for tampering demonstration
import copy

tampered_chain = copy.deepcopy(blockchain.chain)

# Change transaction data in Block 2
if len(tampered_chain) > 1 and tampered_chain[1]['transactions']:
    tampered_chain[1]['transactions'][0]['amount'] = 9999

print("Tampered Transaction:")
print(tampered_chain[1]['transactions'][0])

print("\nChecking tampered blockchain...")

if blockchain.is_chain_valid(tampered_chain):
    print("Blockchain is valid.")
else:
    print("Blockchain is NOT valid because the data was modified.")
````

````python
# Optional Flask code for local execution.
# This cell is mainly for reference in the Jupyter Notebook.

flask_program = r'''
import datetime
import hashlib
import json
from flask import Flask, jsonify, request

class Blockchain:

    def __init__(self):
        self.chain = []
        self.transactions = []
        self.create_block(1, '0', [])

    def create_block(self, proof, previous_hash, transactions):
        block = {
            'index': len(self.chain) + 1,
            'timestamp': str(datetime.datetime.now()),
            'transactions': transactions,
            'proof': proof,
            'previous_hash': previous_hash
        }
        self.chain.append(block)
        return block

    def get_previous_block(self):
        return self.chain[-1]

    def create_transactions(self, sender, receiver, amount):
        transaction = {
            'sender': sender,
            'receiver': receiver,
            'amount': amount
        }
        self.transactions.append(transaction)
        return self.get_previous_block()['index'] + 1

    def proof_of_work(self, previous_proof):
        new_proof = 1
        check_proof = False

        while check_proof is False:
            hash_operation = hashlib.sha256(
                str(new_proof**2 - previous_proof**2).encode()
            ).hexdigest()

            if hash_operation[:3] == '000':
                check_proof = True
            else:
                new_proof += 1

        return new_proof

    def hash(self, block):
        encoded_block = json.dumps(
            block,
            sort_keys=True
        ).encode()

        return hashlib.sha256(encoded_block).hexdigest()

    def is_chain_valid(self, chain):
        previous_block = chain[0]
        block_index = 1

        while block_index < len(chain):
            block = chain[block_index]

            if block['previous_hash'] != self.hash(previous_block):
                return False

            previous_proof = previous_block['proof']
            proof = block['proof']

            hash_operation = hashlib.sha256(
                str(proof**2 - previous_proof**2).encode()
            ).hexdigest()

            if hash_operation[:3] != '000':
                return False

            previous_block = block
            block_index += 1

        return True

app = Flask(__name__)
blockchain = Blockchain()

@app.route('/add_transaction', methods=['POST'])
def add_transaction():
    data = request.get_json()

    required = ['sender', 'receiver', 'amount']

    if not data or not all(key in data for key in required):
        return jsonify({
            'message': 'Missing transaction data'
        }), 400

    index = blockchain.create_transactions(
        data['sender'],
        data['receiver'],
        data['amount']
    )

    return jsonify({
        'message': f'Transaction will be added to Block {index}'
    }), 201

@app.route('/mine_block', methods=['GET'])
def mine_block():

    if len(blockchain.transactions) == 0:
        return jsonify({
            'message': 'No transactions available for mining.'
        }), 400

    previous_block = blockchain.get_previous_block()
    previous_proof = previous_block['proof']

    proof = blockchain.proof_of_work(previous_proof)
    previous_hash = blockchain.hash(previous_block)

    transactions_to_mine = blockchain.transactions.copy()

    # Remove pending transactions before mining
    blockchain.transactions = []

    block = blockchain.create_block(
        proof,
        previous_hash,
        transactions_to_mine
    )

    return jsonify({
        'message': 'Congratulations! You just mined a block!',
        'block': block
    }), 200

@app.route('/get_chain', methods=['GET'])
def get_chain():
    return jsonify({
        'chain': blockchain.chain,
        'length': len(blockchain.chain)
    }), 200

@app.route('/is_valid', methods=['GET'])
def is_valid():
    valid = blockchain.is_chain_valid(blockchain.chain)

    if valid:
        return jsonify({
            'message': 'All good. The Blockchain is valid.'
        }), 200
    else:
        return jsonify({
            'message': 'The Blockchain is not valid.'
        }), 200

if __name__ == '__main__':
    app.run(host='0.0.0.0', port=5000, debug=True)
'''

print("Flask program prepared.")
print("Copy this variable content into blockchain.py for Postman testing.")
````
